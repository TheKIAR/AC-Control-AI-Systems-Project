# Fuzzy Logic Experiments

This notebook contains experiments and results related to the fuzzy logic system implemented in this project. The goal is to evaluate the performance of the fuzzy logic controller for air conditioning.

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path
ROOT = Path.cwd().parent if (Path.cwd() / 'src').exists() else Path.cwd()
for p in (str(ROOT / 'src'), str(ROOT)):
    _ = sys.path.insert(0, p) if p not in sys.path else None
try:
    from fuzzy_logic.fuzzy_system import FuzzySystem
    from fuzzy_logic.rules import create_rules
except ImportError:
    from src.fuzzy_logic.fuzzy_system import FuzzySystem
    from src.fuzzy_logic.rules import create_rules

# Initialize the fuzzy system
fuzzy_system = FuzzySystem()

# Define temperature inputs for testing
temperature_inputs = np.linspace(15, 30, 100)
# evaluate() returns labels; map to levels for plotting
label_to_level = {'Increase Temperature': 1, 'Maintain Temperature': 2, 'Decrease Temperature': 3}
output_labels = []
output_temperatures = []

# Evaluate the fuzzy system for each input temperature
for temp in temperature_inputs:
    output_temp = fuzzy_system.evaluate(float(temp))
    output_labels.append(output_temp)
    output_temperatures.append(label_to_level.get(output_temp, 0))

# Plot the results
plt.figure(figsize=(10, 5))
plt.plot(temperature_inputs, output_temperatures, label='Decision level')
plt.yticks([1, 2, 3], ['Increase', 'Maintain', 'Decrease'])
plt.title('Fuzzy Logic Controller Output')
plt.xlabel('Input Temperature (°C)')
plt.ylabel('Decision')
plt.legend()
plt.grid()
plt.show()
```

## Analysis of Results

The plot above shows the relationship between the input temperature and the output temperature generated by the fuzzy logic controller. Further analysis can be conducted to fine-tune the fuzzy rules and improve the system's performance.
