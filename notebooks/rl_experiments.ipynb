# Reinforcement Learning Experiments

This notebook contains experiments and results related to the reinforcement learning system implemented in this project.

```python
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path
ROOT = Path.cwd().parent if (Path.cwd() / 'src').exists() else Path.cwd()
for p in (str(ROOT / 'src'), str(ROOT)):
    _ = sys.path.insert(0, p) if p not in sys.path else None
try:
    from reinforcement_learning.agent import RLAgent
    from reinforcement_learning.env import RLEnvironment
    from reinforcement_learning.trainer import RLTrainer
except ImportError:
    from src.reinforcement_learning.agent import RLAgent
    from src.reinforcement_learning.env import RLEnvironment
    from src.reinforcement_learning.trainer import RLTrainer

# Initialize environment and agent (env has no action_space attr; use [0, 1])
env = RLEnvironment()
agent = RLAgent(action_space=[0, 1])
trainer = RLTrainer(agent, env)

# Training the agent (train() returns bool; collect rewards manually)
num_episodes = 20
rewards = []
for ep in range(num_episodes):
    s = env.reset()
    done = False
    total = 0
    steps = 0
    while not done and steps < 100:
        prev = s
        a = agent.choose_action(prev, exploration_rate=0.1)
        r, s = env.step(a)
        agent.learn(prev, a, r, s)
        total += r
        done = env.is_terminal_state(s)
        steps += 1
    rewards.append(total)
trainer.is_trained = True

# Plotting the rewards
plt.plot(rewards)
plt.title('Rewards over Episodes')
plt.xlabel('Episode')
plt.ylabel('Reward')
plt.show()

# Evaluate the agent (evaluate() returns a float average)
avg = trainer.evaluate(episodes=5)
print(f'Average Evaluation Reward: {float(avg):.3f}')
```
