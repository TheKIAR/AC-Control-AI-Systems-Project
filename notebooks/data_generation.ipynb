# Data Generation Experiments

This notebook contains experiments and results related to the data generation methods implemented in the project. The focus will be on utilizing the `DataGenerator` class to create synthetic datasets for training and testing the AI systems.

```python
# Import necessary libraries
import numpy as np
import sys
from pathlib import Path
ROOT = Path.cwd().parent if (Path.cwd() / 'src').exists() else Path.cwd()
for p in (str(ROOT / 'src'), str(ROOT)):
    _ = sys.path.insert(0, p) if p not in sys.path else None
try:
    from data_driven.generator import DataGenerator
    from data_driven.pipeline import DataPipeline
    from common.utils import calculate_mean
except ImportError:
    from src.data_driven.generator import DataGenerator
    from src.data_driven.pipeline import DataPipeline
    from src.common.utils import calculate_mean

# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Initialize the DataGenerator
data_generator = DataGenerator()

# Generate synthetic data (returns a list of tuples)
synthetic_data = data_generator.generate_data(num_samples=10)

# Display the first few rows of the generated data
print(synthetic_data[:5])
```

```python
# Clean the generated data
cleaned_data = data_generator.clean_data(synthetic_data)
pipeline = DataPipeline(cleaned_data)
processed = pipeline.process()

# Display the cleaned data
print(processed[:5])
print('mean:', calculate_mean([float(x) for t in processed for x in (t if isinstance(t, (list, tuple)) else [t]) if isinstance(x, (int, float))]))
```

```python
# Save the cleaned data to a CSV file
import csv
out = Path('../data/processed/cleaned_data.csv')
out.parent.mkdir(parents=True, exist_ok=True)
with open(out, 'w', newline='') as f:
    w = csv.writer(f)
    w.writerow(['col0', 'col1', 'col2'])
    w.writerows([r if isinstance(r, (list, tuple)) else [r] for r in processed])
print('Cleaned data saved to ../data/processed/cleaned_data.csv')
```

In this notebook, we successfully generated synthetic data using the `DataGenerator` class, cleaned the data, and saved it for further use in the AI systems. Future experiments will focus on utilizing this data for training and evaluating the fuzzy logic and reinforcement learning models.
